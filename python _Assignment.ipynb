```python
stop_value = int(input("Enter the number that should stop the loop: "))

while True:
    num = int(input("Enter a number: "))

    if num == stop_value:
        print("Stopping the loop.")
        break

    print("You entered:", num)
```

```python
rows = 5

for i in range(rows):
    print(" " * (rows - i - 1), end="")
    print("*" * (2 * i + 1))
```

```python
rows = 5

for i in range(1, rows + 1):
    print(" " * (rows - i), end="")
    for j in range(1, i + 1):
        print("*", end=" ")
    print()
```
