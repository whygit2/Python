Write program to Enter Name and Percentage Marks in a dictionary and display information on screen. 


```python
student = {}
student['Name'] = input("Enter Name: ")
student['Percentage'] = float(input("Enter Percentage Marks: "))

print("\nStudent Information")
for key, value in student.items():
    print(f"{key}: {value}")

print(student)
```

Write a program to find number of occurrence of each letter present in the given string . 


```python
string = input("Enter a string: ")
freq = {}

for char in string:
    if char.isalpha():
        char = char.lower()
        freq[char] = freq.get(char, 0) + 1

print("\n--- Letter Frequency ---")
for letter, count in sorted(freq.items()):
    print(f"{letter}: {count}")
print(freq)    
```

Write a program to find number of occurrence of each vowel present in the given string. 


```python
string = input("Enter a string: ").lower()
vowels = "aeiou"
freq = {v: 0 for v in vowels}

for char in string:
    if char in vowels:
        freq[char] += 1

print("\n--- Vowel Frequency ---")
for vowel, count in freq.items():
    print(f"{vowel}: {count}")
print(freq)    
```

Write a program to accept student names and marks from keyboard and creates as dictionary.
       Also display student marks by taking student name as input. 


```python
students = {}

n = int(input("Enter number of students: "))
for i in range(n):
    name = input(f"Enter name of student {i+1}: ")
    marks = float(input(f"Enter marks of {name}: "))
    students[name] = marks

# Display marks by taking student name as input
search_name = input("\nEnter student name to check marks: ")
if search_name in students:
    print(f"{search_name}'s marks: {students[search_name]}")
else:
    print("Student not found!")

print("\n--- All Student Records ---")
for name, marks in students.items():
    print(f"{name}: {marks}")
```
